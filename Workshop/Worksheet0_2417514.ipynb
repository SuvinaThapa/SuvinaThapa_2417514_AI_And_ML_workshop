```python
#Task2
#mathematical operations on List
def find_sum(numbers):
    """
    Calculate the sum of all number user imput
    parameter is numbers whihc is a list of numeric value
    return float: sum of the given number by user
    """
    return sum(numbers)

def find_average(numbers):
    """
    Calculate the average of all number user imput
    parameter is numbers whihc is a list of numeric value
    return float: avergae of the given number by user
    """
    return sum(numbers) / len(numbers)

def find_max(numbers):
    """
    Caluclate the maximun of all number user imput
    parameter is numbers whihc is a list of numeric value
    return float: max of the given number by user
    """
    return max(numbers)

def find_min(numbers):
    """
    Caluclate the minimum of all number user imput
    parameter is numbers whihc is a list of numeric value
    return float: min of the given number by user
    """
    return min(numbers)

#main program
try:
    user_input = input("Enter numbers separated by space: ")
    numbers = list(map(float, user_input.split()))

    if len(numbers) == 0:
        raise ValueError

    print("Choose operation:")
    print("1 Sum")
    print("2 Average")
    print("3 Maximum")
    print("4 Minimum")

    choice = input("Enter which operation to perform: ")

    if choice == "1":
        print("The total sum of the given number is =", find_sum(numbers))

    elif choice == "2":
        print("The total average of the given number is =", find_average(numbers))

    elif choice == "3":
        print("The maximum from the given number is =", find_max(numbers))

    elif choice == "4":
        print("The minimum from the given number is =", find_min(numbers))

    else:
        print("Invalid operation")

except ValueError:
    print("Invalid input")
```

```python
#4.3 Nested List
#Flatten nested List
def flatten(lst):
    result = []
    for sublist in lst:
        for item in sublist:
            result.append(item)

    return result

print("The required full list from the nested list is :", flatten([[1,2],[3,4],[5]]))


#Accessing the nested eleent
def access_nested_element(lst,indices):
    element = lst
    for i in indices:
        element = element[i]
    return element

lst = [[1,2,3],[4,5,6],[7,8,9]]
print("The required element from the list is:",access_nested_element(lst,[1,2]))

#SUm nisted list
def sum_nested(lst):
    total = 0
    for item in lst:
        if isinstance(item,list):
            total += sum_nested(item)
        else:
            total += item
    return total

print("The required sum of nested list is:",sum_nested([[1,2],[3,[4,5]],6]))


#Remove element
def remove_element(lst,elem):
    result = []
    for item in lst:
        if isinstance(item,list):
            result.append(remove_element(item,elem))
        else:
            if item != elem:
                result.append(item)
    return result
print("The output after removing elemnt 2 from the list is:",(remove_element([[1,2],[3,2],[4,5]],2)))

#Finding Maximum
def find_max(lst):
    max_val = float('-inf')
    for item in lst:
        if isinstance(item,list):
            max_val = max(max_val,find_max(item))
        else:
            max_val = max(max_val,item)
    return max_val
print("The required maximun number from the list is:",find_max([[1,2],[3,[4,5]],6]))

#Count occurance
def count_occurrences(lst,elem):
    count = 0
    for item in lst:
        if isinstance(item,list):
            count += count_occurrences(item,elem)
        else:
            if item == elem:
                count += 1
    return count
print("the required occurance of the number 2 is:",count_occurrences([[1,2],[2,3],[2,4]],2))


#deep flatten
def deep_flatten(lst):
    result = []
    for item in lst:
        if isinstance(item,list):
            result.extend(deep_flatten(item))
        else:
            result.append(item)
    return result
print("The required single list from many nested list is:",(deep_flatten([[[1,2],[3,4]],[[5,6],[7,8]]])))



#Avrage mested
def average_nested(lst):
    flat = deep_flatten(lst)
    return sum(flat)/len(flat)
print("The avergae from the nested list is:",average_nested([[1,2],[3,4],[5,6]]))





```
